# Sequences

A **sequence** is an object that contains other objects in a particular order.  Each item stored in the sequence is called an **element** and each element can be accessed using an integer index that represents the position of the element in the sequence.  There are several build-in sequence types in Python, including:

*   **Lists**: Mutable sequences whose elements can be any type.  List syntax has comma-separated elements enclosed in square brackets `[]`, like `[1, 2, 3]` or `['a', 'b', 'c']`.

*   **Tuples**: Immutable sequences whose elements can be any type.  Tuple syntax has comma-separated elements enclosed in round brackets `()`, like `(1, 2, 3)` or `('a', 'b', 'c')`.

*   **Strings**: Immutable sequences whose elements must be characters.  Strings are created by enclosing characters in quotes, like `'spam'` or `"hamster"`.

Notice that lists are mutable, while tuples and strings are immutable.  **Mutable** means that elements can be changed (or 'mutated') after they have been created.  **Immutable** means can elements not be changed after they have been created.  The important difference between mutable and immutable will become clearer when we learn more about **aliasing** in this lesson.


This lesson will cover each of these sequence types in detail.

### Tuple Assignment

You can put a tuple of variables on the left side of an assignment, and a tuple of values on the right.

```python
a, b = 1, 2
```

The values are assigned to the variables from left to right -- in this example, `a` gets the value `1` and `b` gets the value `2`.
We can display the results like this:

```python
a, b
```

More generally, if the left side of an assignment is a tuple, the right side can be any kind of sequence -- string, list or tuple. 
For example, to split an email address into a user name and a domain, you could write:

```python
email = 'monty@python.org'
username, domain = email.split('@')
```

The return value from `split` is a list with two elements -- the first element is assigned to `username`, the second to `domain`.

```python
username, domain
```

The number of variables on the left and the number of values on the
right have to be the same -- otherwise you get a `ValueError`.

```python
%%expect ValueError
a, b = 1, 2, 3
```

Tuple assignment is useful if you want to swap the values of two variables.
With conventional assignments, you have to use a temporary variable, like this:

```python
temp = a
a = b
b = temp
```

That works, but with tuple assignment we can do the same thing without a temporary variable.

```python
a, b = b, a
```

This works because all of the expressions on the right side are evaluated before any of the assignments.

### Slices

A segment of a sequence is called a **slice**.  Selecting a slice is similar to selecting an element using its index, except that you use `[n:m]` to return the part of the string from the `n`th element to the `m`th element, including the first but excluding the second.

For example, the slice `[3:6]` selects the letters `ana`.  Note that `6` is legal as the second part of a slice, but not legal as an index.


```python
fruit[3:6]
```

If you omit the first index, the slice starts at the beginning of the sequence.

```python
fruit[:4]
```

If you omit the second index, the slice goes to the end of the sequence:

```python
fruit[2:]
```

If the first index is greater than or equal to the second, the result is an **empty sequence**:

## Aliasing

If `a` refers to an object and you assign `b = a`, then both variables refer to the same object.

```python
a = [1, 2, 3]
b = a
b is a
```

So the state diagram looks like this.

```python
t = [1, 2, 3]
binding1 = Binding(Value('a'), Value(repr(t)), dy=-0.11)
binding2 = Binding(Value('b'), draw_value=False, dy=0.11)
frame = Frame([binding1, binding2], dy=-0.25)
```

```python
width, height, x, y = [1.11, 0.81, 0.17, 0.56]
ax = diagram(width, height)
bbox = frame.draw(ax, x, y)
# adjust(x, y, bbox)
```

The association of a variable with an object is called a **reference**.
In this example, there are two references to the same object.

An object with more than one reference has more than one name, so we say the object is **aliased**.
If the aliased object is mutable, changes made with one name affect the other.
In this example, if we change the object `b` refers to, we are also changing the object `a` refers to.

```python
b[0] = 5
a
```

So we would say that `a` "sees" this change, or that changing `b` also changes `a`.
Although this behavior can be useful, it is error-prone.
In general, it is safer to avoid aliasing when you are working with mutable objects.  
Similarly, if you do not expect the sequence to be altered in your code, use an immutable type (tuple or string) instead of a mutable one (list).  

For immutable objects, aliasing is not as much of a problem.
In this example:

```python
a = 'banana'
b = 'banana'
a is b
```

It almost never makes a difference whether `a` and `b` refer to the same
string or not because we can not alter either `a` or `b`.

```python
%%expect TypeError
a[0] = 'T'
print(b)
```

## Lists as function arguments

When you pass a list to a function, the function gets a reference to the
list. If the function modifies the list, the caller sees the change. For
example, `pop_first` uses the list method `pop` to remove the first element from a list.

```python
def pop_first(alist):
    return alist.pop(0)
```

We can use it like this.

```python
letters = ['a', 'b', 'c']
print(f'Before pop_first call letters is {letters}')

first = pop_first(letters)
print(f'After pop_first call letters is {letters}, first is {first}')
```

The return value is the first element, as would be expected from the return statement in `pop_first`.  However, the first element has also been removed from the list -- as we can see by printing the list after the `pop_first` call.  This modification of the `letters` list is possible because lists are mutable.

In this example, the parameter `alist` and the variable `letters` are aliases for the same object, so the state diagram looks like this:

```python
from diagram import Bbox, make_list, adjust, Binding, Value, Frame, Stack, diagram

lst = make_list('abc', dy=-0.3, offsetx=0.1)
binding1 = Binding(Value('letters'), draw_value=False)
frame1 = Frame([binding1], name='__main__', loc='left')

binding2 = Binding(Value('alist'), draw_value=False, dx=0.61, dy=0.35)
frame2 = Frame([binding2], name='pop_first', loc='left', offsetx=0.08)

stack = Stack([frame1, frame2], dx=-0.3, dy=-0.5)
```

```python
width, height, x, y = [2.04, 1.24, 1.06, 0.85]
ax = diagram(width, height)
bbox1 = stack.draw(ax, x, y)
bbox2 = lst.draw(ax, x+0.5, y)
bbox = Bbox.union([bbox1, bbox2])
adjust(x, y, bbox)
```

Passing a reference to an object as an argument to a function creates a form of aliasing.
If the function modifies the object, those changes persist after the function is done.

If you do not want a function to change the value of a sequence parameter then it is better to make the parameter an immutable tuple.

```python
def get_first(atuple):
    return atuple[0]

letters = ('a', 'b', 'c')
print(f'Before get_first call letters is {letters}')

first = get_first(letters)
print(f'After get_first call letters is {letters}, first is {first}')
```

In this example, the `letters` tuple can not be modified by the `get_first` function.

## Argument packing with tuples



Functions can take a variable number of arguments. 
A parameter name that begins with the `*` operator **packs** arguments into a tuple.
For example, the following function takes any number of arguments and computes their arithmetic mean -- that is, their sum divided by the number of arguments. 

```python
def mean(*args):
    return sum(args) / len(args)
```

The parameter can have any name you like, but `args` is conventional.
We can call the function like this:

```python
mean(1, 2, 3)
```

If you have a sequence of values and you want to pass them to a function as multiple arguments, you can use the `*` operator to **unpack** the tuple.
For example, `divmod` takes exactly two arguments -- if you pass a tuple as a parameter, you get an error.

```python
%%expect TypeError
t = (7, 3)
divmod(t)
```

Even though the tuple contains two elements, it counts as a single argument.
But if you unpack the tuple, it is treated as two arguments.

```python
divmod(*t)
```

Packing and unpacking can be useful if you want to adapt the behavior of an existing function.
For example, this function takes any number of arguments, removes the lowest and highest, and computes the mean of the rest.

```python
def trimmed_mean(*args):
    low, high = min_max(args)
    trimmed = list(args)
    trimmed.remove(low)
    trimmed.remove(high)
    return mean(*trimmed)
```

First it uses `min_max` to find the lowest and highest elements.
Then it converts `args` to a list so it can use the `remove` method.
Finally it unpacks the list so the elements are passed to `mean` as separate arguments, rather than as a single list.

Here's an example that shows the effect.

```python
mean(1, 2, 3, 10)
```

```python
trimmed_mean(1, 2, 3, 10)
```

This kind of "trimmed" mean is used in some sports with subjective judging -- like diving and gymnastics -- to reduce the effect of a judge whose score deviates from the others.

```python
character = 'Arthur'
result = []
for line in script:
    if line.startswith(character.upper() + ':'):
        result.append(line)

result
```

Let's encapsulate this in a function.

```python
def lines_by_character(script, character):
    """Return list of lines spoken by `character` (case-insensitive).
    Matches lines that start with the character name in ALL CAPS followed by a colon.
    """
    result = []
    for line in script:
        if line.startswith(character.upper() + ':'):
            result.append(line)
    return result

lines_by_character(script, 'Arthur')
```

Now let's see how many times Arthur says "nee".  Since it might be written as "Nee", or "nee", so we will make the comparison lower case with the `lower()` string method.

```python
for line in lines_by_character(script, 'arthur'):
    if 'nee' in line.lower():
        print(line)
```

And encapsulate that in a function that returns a list.

```python
def words_by_character(script, character, word):
    """
    Return list of lines spoken by `character` (case-insensitive) that contain `word` (case-insensitive).
    """
    result = []
    for line in lines_by_character(script, character):
        if word.lower() in line.lower():
            result.append(line)
    return result

words_by_character(script, 'arthur', 'nee')
```

## Debugging



Note that most list methods modify the argument and return `None`.
This is the opposite of the string methods, which return a new string and leave the original alone.

If you are used to writing string code like this:

```python
word = 'plumage!'
word = word.strip('!')
word
```

It is tempting to write list code like this:

```python
t = [1, 2, 3]
t = t.remove(3)           # WRONG!
```

`remove` modifies the list and returns `None`, so next operation you perform with `t` is likely to fail.

```python
%%expect AttributeError

t.remove(2)
```

This error message takes some explaining.
An **attribute** of an object is a variable or method associated with it.
In this case, the value of `t` is `None`, which is a `NoneType` object, which does not have a attribute named `remove`, so the result is an `AttributeError`.

If you see an error message like this, you should look backward through the program and see if you might have called a list method incorrectly.

## Glossary



**list:**
 An object that contains a sequence of values.

**element:**
 One of the values in a list or other sequence.

**nested list:**
A list that is an element of another list.

**delimiter:**
 A character or string used to indicate where a string should be split.

**equivalent:**
 Having the same value.

**identical:**
 Being the same object (which implies equivalence).

**reference:**
 The association between a variable and its value.

**aliased:**
If there is more than one variable that refers to an object, the object is aliased.

**attribute:**
 One of the named values associated with an object.

 **sequence:**
 An ordered collection of values where each value is identified by an integer index.

**character:**
An element of a string, including letters, numbers, and symbols.

**index:**
 An integer value used to select an item in a sequence, such as a character in a string. In Python indices start from `0`.

**slice:**
 A part of a string specified by a range of indices.

**empty string:**
A string that contains no characters and has length `0`.

**object:**
 Something a variable can refer to. An object has a type and a value.

**immutable:**
If the elements of an object cannot be changed, the object is immutable.

**mutable:**
If the elements of an object can be changed, the object is mutable.

**invocation:**
 An expression -- or part of an expression -- that calls a method.

**pack:**
Collect multiple arguments into a tuple.

**unpack:**
Treat a tuple (or other sequence) as multiple arguments.
