# Lab: Turtle Polygons


Write a program that draws a regular (all sides are equal length) polygon  based on user input consisting of the number of sides, $n$, and the height, $h$, of the polygon.  Here is the math of the problem to help you.  The side length, $s$ of an $n$ sided polynomial (for $n>4$) with height, $h$ is

> $ s = \frac{h}{\tan \theta }$

where $\theta$ is:

> $ \theta = \left( \frac{1}{2}-\frac{1}{n} \right) \pi $

The above is only strictly true for polygons with an even number of sides, but is approximately true for polygons with an odd number of sides.
