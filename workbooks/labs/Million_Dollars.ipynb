## Write some small helper functions

The following functions are designed to *decompose* the problem into a collection of *abstract* functions with *low coupling* that can be put together to solve the larger problem in the `main` code.  These functions are designed to solved the problem by *iterating* over each month and calculating the amount of savings *accumulated* over your career.

- **`monthly_interest(savings, annual_interest_rate)`** Returns the amount of interest accrued for the month on the current savings amount.  Example; if the savings account has \$100,000 and the annual interest rate is 12%, then the function will return \$1,000.



- **`is_anniversary(month_number)`** Returns True if the number of months since starting work (`month_number`) is an anniversary and thus you should get a raise.  Example; if you are in the 15th month of work the function should return False, if you are in your 36th month of work the function should return True.  Note that the first month of work should not considered an anniversary because you don't (typically) get a raise in your first month.

- **`apply_raise(salary, annual_raise)`** Returns the new salary after the annual raise has been applied to the current salary.  Example; a salary of \$1,000 with a raise of 3% the function should return \$1,030
