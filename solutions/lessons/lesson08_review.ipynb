## Functions
A function is a block of organized, reusable code that performs a specific task. A well designed function should do one thing well.

### Why use Functions?

 * **Modularity**: Functions break down a large program into smaller, manageable, and organized pieces. This makes the code easier to understand, debug, and maintain.

* **Reusability**: Once a function is defined, it can be called multiple times from different parts of the program without rewriting the same code. This promotes code efficiency and reduces redundancy.

* **Abstraction**: Functions allow you to hide the complex implementation details from the main program. You only need to know what a function does, not how it does it.

### Key Concepts:
* **Definition**: This is where you write the code for the function, specifying its name, parameters, and the operations it performs.

* **Parameters**: These are variables listed inside the parentheses of the function definition. They act as placeholders for the values that the function needs to operate on when it's called.

* **Docstring** A description that explains what the function does.

* **Return Value**: Functions often compute a result and 'return' that value to the part of the code that called it.  Some functions do not have a return value.

* **Calling a Function**: To execute the code inside a function, you 'call' the function often passing in actual values (arguments) for its parameters.

* **Scope**: This refers to the region of the program where a variable is accessible. Variables defined inside a function are typically local to that function and cannot be accessed from outside.

* **Unit Tests** Code used to validate the function operates correctly, simple unit tests use 'assert' statements.


In essence, functions are like mini-programs within your main program, each with a specific job, making your code more structured and efficient.

Here is an example of a function that extracts the odd numbers from a list of numbers. 

```python
def extract_odd_numbers(numbers_list):
    """
    Returns all odd numbers from a given list of integers.
    Args:
        numbers_list (list): A list of integers.
    Returns:
        list: A new list containing only the odd numbers from the input list.
    """
    odd_numbers = []
    for num in numbers_list:
        # Ensure the item is an integer before checking for odd/even
        if num % 2 != 0:
            odd_numbers.append(num)
    return odd_numbers

# unit tests
assert extract_odd_numbers([1, 2, 3, 4, 5]) == [1, 3, 5]
assert extract_odd_numbers([2, 4, 6, 8]) == []
assert extract_odd_numbers([0, -1, -2, -3]) == [-1, -3]
assert extract_odd_numbers([21, 12, 42, 91, 52]) == [21, 91]
```

## Conditionals
Conditionals (`if`, `elif`, `else`) allow a program to make decisions and execute different blocks of code based on whether a certain condition is true or false. 

### Why use Conditionals?

 * **Decision Making**: Conditionals enable programs to perform different actions based on different inputs or states. 

 * **Controlling Flow**: Dictate the order in which instructions are executed, allowing for non-linear execution paths that adapt to circumstances.

### Key Concepts:

 * **`if` statement**: The most basic conditional. The code block following if executes only if its condition is true.

 * **`else` statement**: Used in conjunction with if. The code block following else executes if the if condition (and any elif conditions) is false.

 * **`elif` statement**: Allows you to check multiple conditions sequentially. If the if condition is false, the program checks the elif conditions in order until one is true, or falls back to else if none are true.

 * **Conditions/Expressions**: These are statements that evaluate to either True or False. They often involve comparison operators (==, !=, >, <, >=, <=) and logical operators (and, or, not).

 * **Indentation**: Indentation defines which lines of code belong to which if, elif, or else block.
 
In essence, conditionals give your programs the ability to think and react, making them capable of solving more complex and real-world problems by navigating various possibilities.

Here is a example of a conditional chain.

```python
weather = "sunny"
temperature = 25

if weather == "sunny":
    if temperature > 25:
        outfit = "shorts and t-shirt"
    elif temperature > 15:
        outfit = "jeans and t-shirt"
    else:
        outfit = "long pants and sweater"
elif weather == "rainy":
    outfit = "raincoat and boots"
else:
    outfit = "coat and hat"

print(f"Recommended outfit: {outfit}")
```

## Iteration

Iteration (also called looping) allows a program to execute a block of code repeatedly, either a fixed number of times (a `for` loop) or until a certain condition is met (a `while` loop). Loops enable you to process multiple items efficiently without writing repetitive code.

### Why use Iteration?

* **Efficiency**: Instead of writing the same code multiple times, you write it once in a loop and let it execute repeatedly. This saves time and reduces code length.

* **Processing Collections**: Iteration allows you to process each item in a list, tuple, string, or other collection without knowing the collection's size in advance.

* **Automation**: Loops automate repetitive tasks, such as performing calculations, transforming data, or checking conditions on multiple items.

### Key Concepts:

* **`for` loop**: Iterates over a sequence (list, tuple, string, range, etc.) and executes a block of code once for each item. The loop variable takes on the value of each item in sequence.

* **`while` loop**: Continues to execute a block of code as long as a specified condition is true. Useful when you don't know in advance how many times the loop needs to run.

* **Loop Variable**: The variable in a `for` loop that takes on each value from the sequence. In `while` loops, a variable is typically checked and updated within the loop.

* **Iteration Range**: The `range()` function generates a sequence of numbers, commonly used in `for` loops to iterate a specific number of times.

* **Break Statement**: Prematurely exits a loop before the normal loop condition is met.

* **Continue Statement**: Skips the remaining code in the current iteration and jumps to the next iteration of the loop.

* **Nested Loops**: A loop inside another loop, useful for processing multi-dimensional data or combinations of items.

In essence, loops are fundamental tools that let your programs handle repetitive tasks automatically, making code more concise, maintainable, and capable of handling varying amounts of data.

```python
# For loops

# a list of fruits
fruits = ["apple", "banana", "cherry", "pinapple"]

# loop over items in the fruit list
for fruit in fruits:
    print(f"I like {fruit}s")

# loop over the indexes of the fruit list
for i in range(len(fruits)):
    print(f"Fruit {i}: {fruits[i]}")
```

```python
# While loop example: 

# initialize a countdown variable
countdown = 3

# repeat until a condition becomes false
while countdown > 0:
    print(f"  {countdown}...")
    countdown -= 1

print("Blastoff!")
```

## Sequences

Sequences are ordered collections of items in Python, such as lists, tuples, and strings. They allow you to store multiple values together and access individual elements by their position (index) within the collection.

### Why use Sequences?

* **Grouping Related Data**: Sequences let you store multiple related values in a single variable rather than creating separate variables for each value.

* **Processing Multiple Items**: Combined with iteration, sequences allow you to efficiently process many items with a single loop, rather than writing code for each individual item.

* **Indexed Access**: Each element in a sequence has a position (index), allowing you to access specific items directly without iterating through the entire sequence.

* **Flexibility**: Different sequence types (lists, tuples, strings) offer different capabilities—lists are mutable (changeable), while tuples and strings are immutable (fixed once created).

### Key Concepts:

* **List**: A mutable (changeable) ordered collection of items enclosed in square brackets `[]`. Lists allow you to add, remove, or modify elements.

* **Tuple**: An immutable (unchangeable) ordered collection of items enclosed in parentheses `()`. Once created, tuples cannot be modified, making them useful for protecting data.

* **String**: A sequence of characters (text). Strings behave like sequences and can be indexed and iterated over, but each character is immutable.

* **Indexing**: Accessing individual elements using their position in the sequence. Python uses zero-based indexing, where the first element is at index 0, the second at index 1, and so on.

* **Slicing**: Extracting a portion of a sequence using the syntax `sequence[start:end]`, which returns elements from the start index up to (but not including) the end index.

* **Length**: The `len()` function returns the number of items in a sequence, useful for understanding the size of your collection.

* **Membership Testing**: Using the `in` operator to check whether a value exists in a sequence (e.g., `"apple" in fruits` returns True or False).

* **Mutability**: Lists are mutable and can be modified in place; tuples and strings are immutable and cannot be changed after creation.

In essence, sequences are fundamental data structures that enable you to organize, access, and manipulate collections of data efficiently, forming the foundation for storing and processing real-world information in your programs.

```python
# Sequences example: Working with lists, tuples, and strings

# Create a list of desserts
desserts = ["cake", "ice cream", "chocolate", "pie"]

# Indexing: Access individual elements by position (zero-based)
print("First dessert:", desserts[0])
print("Third dessert:", desserts[2])

# Using len() to find the number of items
print(f"Number of desserts: {len(desserts)}")

# Slicing: Extract a portion of the list
print("First two desserts:", desserts[0:2])
print("Last two desserts:", desserts[2:4])

# Membership testing: Check if an item is in the sequence
if "cake" in desserts:
    print("Cake is in the list")

# Iterating over a sequence
print("\nAll desserts:")
for dessert in desserts:
    print(f"  - {dessert}")

# Lists are mutable: you can modify elements
desserts[1] = "donut"
print(f"\nAfter modifying: {desserts}")

# Tuples are immutable
coordinates = (10, 20, 30)
print(f"\nCoordinates tuple: {coordinates}")
print(f"First coordinate: {coordinates[0]}")

# Strings are sequences of characters
message = "hello"
print(f"\nMessage: {message}")
print(f"First letter: {message[0]}")
print(f"Letters 1-3: {message[1:4]}")
```

## Dictionaries

Dictionaries are unordered collections of key:value pairs in Python, enclosed in curly braces `{key:value, key:value, ...}` or created with the `dict()` constructor. They allow you to store and retrieve data using meaningful keys instead of numeric indices, making them ideal for representing structured, labeled information.

### Why use Dictionaries?

* **Named Access**: Instead of remembering numeric indices, you access values using descriptive keys (e.g., `person["name"]` instead of `person[0]`), making code more readable and self-documenting.

* **Fast Lookups**: Dictionaries provide efficient key-based lookups, allowing you to retrieve values quickly without iterating through a collection.

* **Flexibility**: Keys can be strings, numbers, or other immutable types, allowing diverse ways to organize and access your data.

### Key Concepts:

* **Key-Value Pair**: A dictionary stores data as pairs where each key is associated with a value. Keys must be unique and immutable; values can be any type.

* **Accessing Values**: Retrieve a value using its key with square brackets, e.g., `person["name"]` returns the value associated with the key "name".

* **Adding/Modifying Entries**: You can add new key-value pairs or modify existing values by assigning to a key, e.g., `person["age"] = 31`.

* **Dictionary Methods**: Useful methods include `.keys()` (returns all keys), `.values()` (returns all values), `.items()` (returns key-value pairs), and `.get()` (safely retrieves a value with a default if the key doesn't exist).

* **Iteration**: You can iterate over a dictionary's keys, values, or key-value pairs using `for` loops.

* **Membership Testing**: Use the `in` operator to check whether a key exists in a dictionary (e.g., `"name" in person`).

In essence, dictionaries are powerful data structures for mapping meaningful labels to values, enabling you to write more expressive and maintainable code when working with structured data where named access is more important than positional access.

```python
country_capitals = dict(Canada="Ottawa", England="London", Germany="Berlin")
country_capitals["India"] = "New Delhi"
country_capitals.update({"Japan": "Tokyo", "Australia": "Canberra"})

# Loop over the dictionary
for key, value in country_capitals.items():
    print(f"{key}: {value}")
```

## Sets

Sets are unordered collections of unique items in Python, enclosed in curly braces `{item1, item2, ...}` or created using the `set()` constructor. They are mutable, meaning you can add or remove elements, but they do not allow duplicate values and do not support indexing or slicing like lists or tuples.

### Why use Sets?

* **Uniqueness**: Sets automatically eliminate duplicates, ensuring each element appears only once. This is useful for tasks like removing duplicates from a list or checking membership efficiently.

* **Fast Membership Testing**: Checking if an item is in a set is very fast, making sets ideal for operations like filtering or finding intersections.

* **Mathematical Operations**: Sets support operations like union, intersection, difference, which are useful for comparing collections of data.

### Key Concepts:

* **Creating Sets**: Use curly braces `{1, 2, 3}` for non-empty sets or `set()` for an empty set. Note that `{}` creates an empty dictionary, not a set.

* **Adding Elements**: Use `.add(item)` to add a single item or `.update(iterable)` to add multiple items from a list, tuple, or another set.

* **Removing Elements**: Use `.remove(item)` to remove an item (raises an error if not found) or `.discard(item)` (does nothing if not found). 

* **Set Operations**:
    - **Union**: `set1 | set2` or `set1.union(set2)` – combines all unique elements from both sets.
    - **Intersection**: `set1 & set2` or `set1.intersection(set2)` – elements common to both sets.
    - **Difference**: `set1 - set2` or `set1.difference(set2)` – elements in set1 but not in set2.

* **Membership Testing**: Use the `in` operator to check if an item exists in the set (e.g., `1 in my_set`).

* **Iteration**: You can loop over a set using a `for` loop, but the order is not guaranteed.

In essence, sets are useful for handling collections where uniqueness and fast lookups are important, enabling efficient data processing and mathematical set operations in your programs.

```python
fruits = ["apple", "banana", "kiwi", "pinapple"]
desserts = ["cake", "chocolate", "pie", "apple", "pinapple"]

# Create sets from fruits and desserts
fruit_set = set(fruits)
dessert_set = set(desserts)

# Union: combine all unique elements
all_items = fruit_set | dessert_set
print("Union of fruits and desserts:", all_items)

# Intersection: common elements
common_items = fruit_set & dessert_set
print("Intersection:", common_items)

# Difference: elements in fruit_set but not in dessert_set
unique_fruits = fruit_set - dessert_set
print("Fruits not in desserts:", unique_fruits)

# Adding elements
fruit_set.add("orange")
print("After adding 'orange' to fruits:", fruit_set)

# Membership testing
print("'apple' in fruits:", "apple" in fruit_set)
```

## File Input/Output

File Input/Output (I/O) is how programs read data from and write data to files. It lets you keep information between runs of your program, work with larger datasets, and interact with real-world data stored in files.

### Why use File I/O?

* **Persistence**: Save program output to a file so it can be used later, or load existing data into a program without hard-coding values.

* **Data Exchange**: Files are a common way for programs to share data with each other or with users (e.g., logs, data tables, configuration files).


### Key Concepts:

* **Opening Files**: Use `open(filename, mode)` to get a file object. Common modes include:
  * `'r'` — read (default)
  * `'w'` — write (overwrites existing file)

* **Context Manager (`with` statement)**: Use `with open(...) as f:` to ensure the file is automatically closed when done, even if an error occurs.

* **Reading**: Common methods include:
  * `f.read()` — read the entire file as a string
  * `f.readline()` — read the next line
  * `f.readlines()` — read all lines into a list

* **Writing**: Use `f.write(text)` to write strings to a file. Remember that writing does not automatically add newlines.

In essence, file I/O is a key bridge between your program and the outside world, letting your code persist results and work with real datasets.

```python
text_to_save = "We are the knights who say Ni!.\nWe want... a shrubbery!\n"

with open("quote.txt", "w") as f:
    f.write(text_to_save)

with open("quote.txt", "r") as f:
    loaded_text = f.read()

print(loaded_text)
```

## Classes

Classes let you define new types that bundle together data (attributes) and behavior (methods) into a single reusable structure.

### Why use Classes?

* **Model real-world things**: Represent entities like points, students, or game characters with attributes and actions.

* **Encapsulation**: Group related data and functions together, keeping internal details hidden and exposing a clean interface.

* **Reuse and organization**: Define logic once in a class and create many instances (objects) that share that behavior.

### Key Concepts:

* **Class Definition**: Use the `class` keyword, e.g., `class Point:`.

* **Instance**: An object created from a class (e.g., `p = Point(3, 4)`).

* **`__init__` method**: The initializer (constructor) that sets up new instances.

* **Attributes**: Variables attached to an instance, accessed with `self.attribute` inside methods and `instance.attribute` outside.

* **Methods**: Functions defined inside a class that operate on instances (e.g., `move`, `draw`).

* **`self`**: The first parameter in instance methods, referring to the current object.

* **Dunder methods**: Special methods to define how objects behave with Python's built-in functions and operators.

In essence, classes are the foundation of object-oriented programming in Python, helping you build structured, maintainable code for complex programs.

```python
import random

# Define a Card class to represent a playing card
class Card:
    def __init__(self, suit, rank):
        self.suit = suit
        self.rank = rank
    
    def __str__(self):
        return f"{self.rank} of {self.suit}"

# Define a Deck class to represent a deck of cards
class Deck:
    def __init__(self):
        self.cards = []
        for suit in ["Hearts", "Diamonds", "Clubs", "Spades"]:
            for rank in ["Ace", "2", "3", "4", "5", "6", "7", "8", "9", "10", "Jack", "Queen", "King"]:
                self.cards.append(Card(suit, rank))
    
    def shuffle(self):
        random.shuffle(self.cards)
    
    def deal(self):
        return self.cards.pop()
    
    def __str__(self):
        return f"Deck with {len(self.cards)} cards"

# Create a deck
deck = Deck()
print(deck)

# Shuffle the deck
deck.shuffle()

# Deal a card
card = deck.deal()
print(f"Dealt: {card}")
```

## Algorithms

Throughout the course we have seen several algorithms that can be used to solve a broad range of problems. 

- **Counter Algorithms**: Counts the number of elements that satisfy a condition within a sequence. Initialize a counter to 0, loop over the sequence, increment the counter for each element that satifies the condition. Example: counting the number of 'a's in a string.  Dictionaries can be used to count elements that satisfy different conditions in a single loop through the sequence. Example: counting the number of each letter that appears in a string.

```python
def counter_algorithm(input_list):
    """
    Counts the occurrences of each unique item in the input list.
    Args:
        input_list (list): A list of items to count.
    Returns:
        dict: A dictionary where keys are unique items and values are their counts.
    """
    
    # Initialize an empty dictionary to store counts
    count_dict = {}
    
    # Loop through each item in the input list
    for item in input_list:
        
        # if the item is already in the dictionary, increment its count
        if item in count_dict:
            count_dict[item] += 1
        # if the item is not in the dictionary, add it with a count of 1
        else:
            count_dict[item] = 1
    
    # Return the dictionary with counts
    return count_dict

assert counter_algorithm(["apple", "banana", "apple", "orange", "banana", "apple"]) == {"apple": 3, "banana": 2, "orange": 1}
assert counter_algorithm([]) == {}
assert counter_algorithm(["a", "b", "c", "a", "b", "a"]) == {"a": 3, "b": 2, "c": 1}
assert counter_algorithm([1, 2, 3, 2, 1, 1]) == {1: 3, 2: 2, 3: 1}
```

- **Accumulator Algorithms**: Builds up (accumulates) a result by iteratively combining elements from a sequence with an initial value. Initialize an accumulator variable (e.g., to 0 for summing, empty list for collecting, etc.), loop over the sequence, update the accumulator with each element. Examples: summing numbers in a list, concatenating strings, finding the maximum value. Dictionaries can accumulate different results in a single pass.

```python
def above_threshold(input_list, threshold):
    """
    Returns a list of all numbers in the input list that are greater than a specified threshold.
    Args:
        input_list (list): A list of numbers to filter.
        threshold (number): The threshold value to compare against.
    Returns:
        list: A list of numbers greater than the threshold.
    """
    
    # initialize an empty list to store the results
    result = []
    
    # loop through each number in the list and check if it's greater than the threshold
    for num in input_list:
        if num > threshold:
            result.append(num)
            
    # return the list of numbers above the threshold
    return result

assert above_threshold([1, 5, 3, 8, 2], 4) == [5, 8]
assert above_threshold([10, 20, 30, 40], 25) == [30, 40]
assert above_threshold([-1, -5, -3, -8, -2], -4) == [-1, -3, -2]
```

- **Linear Search Algorithm**: Searches for a specific value in a sequence by iterating through each element until found or end reached. Returns True (or the index of the element) if found, False (or -1) otherwise.  Example: determining if a word spelled backwards is a valid word.

```python
def linear_search(input_list, target):
    """
    Performs a linear search for a target value in the input list.
    Args:
        input_list (list): A list of items to search through.
        target: The value to search for.
    Returns:
        bool: True if the target is found in the list, False otherwise."""
        
    # loop through each item in the list and check if it matches the target
    for item in input_list:
        if item == target:
            return True
    
    # if you get here, the target was not found in the list   
    return False
```

- **Recursive Algorithm**: Solves a problem by writing a function to solve the simplest instance of the problem then repeatedly (recursively) calling the function until some stopping condition (the base case) is met.  Example: factorial.

```python
def factorial(n):
    """
    Computes the factorial of a non-negative integer n.
    Args:
        n (int): A non-negative integer whose factorial is to be computed.
    Returns:
        int: The factorial of n, denoted as n!.
    """
    
    # Base case: factorial of 0 is 1
    if n == 0:
        return 1
    
    # Recursive case: n! = n * (n-1)!
    return n * factorial(n - 1)
```

 - **Search Algorithm Efficiency**: This is not really an algorithm, but a reminder that our choice of algorithm and tool (list, dictionary, set) can impact how long it takes for the algorithm to run.  Example from the reversible words search:
  - *Slow Algorithm*: For each word in the list, use a nested loop to check if its reverse is also in the list. O(n²) time complexity.
  - *Fast Algorithm*: For each word in the list, use a dictionary to search for the reversed word.  O(n) time complexity due to dictionaries being hash tables.


```python
def search_sum_slow(input_list, target):
    """
    Returns True if any two items in the input list sum to the target value, and False otherwise.
    Args:
        input_list (list): A list of items to search through.
        target: The value to search for.
    Returns:
        bool: True if the target is found in the list, False otherwise."""
    
    # loop through each item in the list
    for i in range(len(input_list)):
        
        # pair it with every other item in the list
        for j in range(i + 1, len(input_list)):
        
            # check if their sum equals the target
            if input_list[i] + input_list[j] == target:
                
                # if we find a pair that sums to the target, return True
                return True
    
    # if you get here, no two items sum to the target
    return False




def search_sum_fast(input_list, target):
    """
    Returns True if any two items in the input list sum to the target value, and False otherwise.
    Args:
        input_list (list): A list of items to search through.
        target: The value to search for.
    Returns:
        bool: True if the target is found in the list, False otherwise."""

    # Create a dictionary of the numbers minus the target
    num_dict = {}
    for num in input_list:
        num_dict[num] = target - num
    
    # Loop through the list again and check if the complement is in the dictionary
    for num in input_list:
        complement = target - num
        if complement in num_dict and complement != num:
            return True
           
    # If you get here, no two items sum to the target
    return False

# make a big random list of integers for testing
big_random_list = []
for _ in range(1000):
    big_random_list.append(random.randint(1, 1000))

# test the search_sum_slow and search_sum_fast functions and time them
print("Testing search_sum_slow...")
%timeit search_sum_slow(big_random_list, 8362)

print("Testing search_sum_fast...")
%timeit search_sum_fast(big_random_list, 8362)

```

## Design, Develop, and Debug Strategies 

### Design Strategies

Before we write our algorithms into programs we make program design choices.  Here are some good programing design concepts:

- **Decomposition**: Break down large or complex problems into smaller, manageable, and independent parts.  Validating that these independent parts each work independently before bringing them together to solve the larger problem will (likely) reduce debugging time.

- **Encapsulation**: Enclose complexities within functions, methods, and classes.  Functions and methods should do small things so that they are easy to test.  This makes it easier to test your code and inceases readability.  Use pure functions (functions that don't modify parameters and only return values) whenever reasonable, as they are less error-prone.

- **Generalization**: The process of replacing something unnecessarily specific (like a constant value) with something general (like a variable or parameter). Generalization makes code more versatile, more likely to be reused, and sometimes even easier to write.  If you find yourself writing the same line of code more than a few times, it is likely time to encapsulate and generalize that line of code.


### Development Strategies

When we are in the process of writing (developing) code, we need to decide how to write and test the code efficiently.  Here are some development strategies:

- **Incremental Development**: Add and test only a small amount of code at a time to reduce time spent debugging.
  1. Start with a working program, make small changes, and test after every change.
  2. Add *scaffolding* (e.g., temporary print statements for debugging) to display and check the values of variables.
  3. Once the program is working, remove the scaffolding.

- **Prototype and Patch**: Start with a rough draft and gradually add features and fix bugs as you find them.  Build testing as you go and add features as 'patches' to your existing rough draft.  Prototype and patch can be effective when you don't have a deep understanding of the problem, but can lead to complicated code.  It is best used as a first step toward design-first development.

- **Design-First Development**: Involves more planning before prototyping. Use high-level insights into the problem to make programming easier (e.g., thinking of Time objects as base-60 numbers).  This strategy requires some familiarity with what code you are able to write, so it takes some time to build the experience to do effective design-first development.  For beginner (and intermediate) programmers, design-first development often comes after prototype and patch - once you have cobbled together a working version, you see a better way to do it, so you scrap your code and start again - but with a clear design to start from.

### Debug Strategies

*"If debugging is the process of removing software bugs, then programming must be the process of putting them in."* (Edsger W. Dijkstra) Remember that debugging is a normal part of the programming cycle.  Accept that it can be frustrating and that even great programmers write bugs.  Good programmers focus on avoiding bugs, great programmers focus on finding bugs. 

- **Understand Error Messages**: Understand types of errors - syntax (code structure), runtime (execution errors), semantic (incorrect logic). 

- **Scaffolding**: Use temporary print statements to inspect intermediate values, then remove once working.

- **Unit Testing**: Use `assert` statements to validate function behavior early - they raise `AssertionError` if conditions fail, helping catch issues during testing.

- **Sample Testing**: For large datasets, check your code on a small sample of the data before running on the whole set.  Print data types to check mismatches, perform sanity checks (e.g., results within expected ranges), consistency checks (compare different computations), and unit testing on a small (well understood) part of your data.

### Reversible Primes
A number that is only divisible by two whole numbers (one and itself) is called a *prime number*.  For example, 11 is a prime number because it is only divisible by 1 and 11.  On the other hand, 14 is not a prime number because it is divisible by 1, 2, 7, and 14.  Note that 1 is not a prime number because it is only divisible by one whole number, 1.
A number that reads the same forward and backward is called a *reversible number*.  For example 101, 23232, and 9889 are reversible numbers.

 - Write a boolean function that takes a single integer parameter and returns True iff the number is a prime number. 
 - Write a second boolean function that takes a single integer parameter and returns True iff the number is reversible.
 - Write a third function that takes a single integer parameter and returns the sum of all reversible prime numbers from zero up to (and including) that number.

```python
# Solution

def is_prime(n):
    """Return True if n is a prime number."""
    
    # check that n is not 0 or 1, since those are not prime
    if n < 2:
        return False
    
    # check for factors from 2 up to n/2 
    # if n is divisible by any number in this range, it's not prime
    for i in range(2, n//2 + 1):
        if n % i == 0:
            return False
    
    # if we get here, n is prime    
    return True


def is_reversible(n):
    """Return True if n reads the same forward and backward."""

    # convert the number to a string and 
    # compare characters from the start and end
    digits = str(n)

    # loop through the first half of the digits and 
    # compare to the corresponding digit from the end
    # if any pair of digits doesn't match, return False    
    for i in range(len(digits)//2):
        if digits[i] != digits[-(i + 1)]:
            return False
    
    # if we get here, the number is reversible    
    return True


def reversible_prime_sum(upper_limit):
    """Return the sum of all reversible prime numbers below upper_limit."""
    
    # initialize a total variable to keep track of the sum
    total = 0
    
    # loop through numbers starting from 2 up to upper_limit
    for num in range(2, upper_limit):
        if is_prime(num) and is_reversible(num):
            # print(f'{num:6}') # this is scaffolding that can be removed later
            total += num
    
    # return the final total after the loop is done
    return total


# unit tests
assert reversible_prime_sum(100) == 28  # 2, 3, 5, 7, 11 are reversible primes below 100
assert reversible_prime_sum(200) == 783  # includes 101, 131, 151, 181, 191
assert reversible_prime_sum(10) == 17  # 2, 3, 5, 7 are reversible primes below 10
assert reversible_prime_sum(1) == 0  # no primes below 1
```

### Scrabble Score 
Scrabble is a board game in which players spell words from a group of letters in their hand.  The players hand has 7 letters and is connected to the board by a letter on the board, so words are between 2 and 8 letters (inclusive).  The score of their word is the sum of the individual word scores.  The SCRABBLE_SCORES dictionary below gives the score for each letter in Scrabble.  The SCRABBLE_TILES dictionary below gives the total number of tiles available for each letter.

 - Write a function that checks that a word can be spelled from a Scrabble hand.  It must contain 8 or less letters and must not contain more of any single letter than is available in the game (ie. 'pizza' is not valid because it contains two z's, but there is only one z in the game).  
 - Write a function that returns the scrabble score of a word.  
 - Make a list of the highest 10 scrabble score words in the English language using the `words.txt` file.

```python
SCRABBLE_SCORES = {
    'a': 1, 'b': 3, 'c': 3, 'd': 2,
    'e': 1, 'f': 4, 'g': 2, 'h': 4,
    'i': 1, 'j': 8, 'k': 5, 'l': 1,
    'm': 3, 'n': 1, 'o': 1, 'p': 3,
    'q': 10, 'r': 1, 's': 1, 't': 1,
    'u': 1, 'v': 4, 'w': 4, 'x': 8,
    'y': 4, 'z': 10
}

SCRABBLE_TILES = {
    'a': 9,  'b': 2,  'c': 2,  'd': 4,  'e': 12,
    'f': 2,  'g': 3,  'h': 2,  'i': 9,  'j': 1,
    'k': 1,  'l': 4,  'm': 2,  'n': 6,  'o': 8,
    'p': 2,  'q': 1,  'r': 6,  's': 4,  't': 6,
    'u': 4,  'v': 2,  'w': 2,  'x': 1,  'y': 2,
    'z': 1
}
```

```python
# Solution

def is_valid_scrabble_word(word):
    """Return True if the word can be made with the available Scrabble tiles."""
    
    # ensure word is between 2 and 8 letters long
    if len(word) < 2 or len(word) > 8:
        return False
    
    # create a dictionary to count the frequency of each letter in the word
    letter_count = {}
    
    # loop through each character in the word
    for ch in word.lower():
        if ch in SCRABBLE_SCORES:  # only consider valid letters
            letter_count[ch] = letter_count.get(ch, 0) + 1
    
    # check if the frequency of each letter exceeds the available tiles
    for letter, count in letter_count.items():
        if count > SCRABBLE_TILES.get(letter, 0):
            return False
    
    return True


def scrabble_score(word):
    """Return the Scrabble score of a word."""
    score = 0
    for ch in word.lower():
        if ch in SCRABBLE_SCORES:
            score += SCRABBLE_SCORES[ch]
    return score

    
def top_scrabble_words(filename):
    """Return the top 10 highest scoring words from a file."""
    
    # create an empty list to store (word, score) pairs
    words_scores = []
    
    # open the file and read it line by line
    with open(filename, "r") as file:
        
        # loop through each line in the file
        for line in file:
            
            # remove leading/trailing whitespace
            word = line.strip()
            
            if is_valid_scrabble_word(word):
                score = scrabble_score(word)
                words_scores.append((score, word))
    
    words_scores.sort(reverse=True)
    
    return words_scores[:10]

print(top_scrabble_words("words.txt"))
```

### Word Frequency 
 - Write a function that removes any punctuation characters from a text.  You can get a string containing the punctuation characters using `string.punctuation` from the `string` module.
 - Read `holygrail.txt` and write a function `word_frequencies(filename)` that returns a dictionary mapping each word to how many times it appears in the file.  Exclude the names of characters.  
 - Sort using the values of the dictionary to print the words from most common to least common, excluding the 'STOP_WORDS' given in the set below. 

```python
STOP_WORDS = {
    "a","about","above","after","again","against","all","am","an","and","any","are","aren't",
    "as","at","be","because","been","before","being","below","between","both","but","by",
    "can","can't","cannot","could","couldn't","did","didn't","do","does","doesn't","doing","don't",
    "down","during","each","few","for","from","further","had","hadn't","has","hasn't","have","haven't",
    "having","he","he'd","he'll","he's","her","here","here's","hers","herself","him","himself","his",
    "how","how's","i","i'd","i'll","i'm","i've","if","in","into","is","isn't","it","it's","its","itself",
    "let's","me","more","most","mustn't","my","myself","no","nor","not","of","off","on","once","only",
    "or","other","ought","our","ours","ourselves","out","over","own","same","she","she'd","she'll",
    "she's","should","shouldn't","so","some","such","than","that","that's","the","their","theirs",
    "them","themselves","then","there","there's","these","they","they'd","they'll","they're","they've",
    "this","those","through","to","too","under","until","up","very","was","wasn't","we","we'd","we'll",
    "we're","we've","were","weren't","what","what's","when","when's","where","where's","which","while",
    "who","who's","whom","why","why's","with","won't","would","wouldn't","you","you'd","you'll","you're",
    "you've","your","yours","yourself","yourselves"
}
```

```python
# Solution

import string 

def remove_punctuation(text):
    result = ""
    for ch in text:
        if ch == "'" or ch not in string.punctuation:
            result += ch
    return result


def word_frequencies(filename):
    """Return a dictionary mapping words to their frequency,
       excluding character names."""
    
    # initialize an empty dictionary to store word frequencies
    freq = {}
    
    # open the file for reading
    with open(filename, "r") as file:
        
        # read the file line by line
        for line in file:
            
            # Remove leading/trailing whitespace
            line = line.strip()
            
            # Skip empty lines
            if not line:
                continue
            
            # Skip lines that don't follow the expected format CHARACTER: dialogue
            if len(line.split(': ')) < 2:
                continue
            
            # eliminate the character name that appears at start of line (e.g., "ARTHUR:" or "KING ARTHUR:")
            line = line.split(': ')[-1]
            
            # make the line lowercase
            line = line.lower()
            
            # remove punctuation from the line
            line = remove_punctuation(line)
            
            # convert the line into a list of words
            words = line.split()

            # loop through the words in the line,
            # check they are not in STOP_WORDS, and 
            # update their frequency in the dictionary            
            for word in words:
                if word in STOP_WORDS:
                    continue
                if freq.get(word, 0) == 0:
                    freq[word] = 1
                else:
                    freq[word] += 1
                    
    # after processing all lines, return the frequency dictionary
    return freq


def sort_dictionary_by_value(word_frequencies):
    """Return a list of (key, value) pairs sorted by value in descending order."""
    counter = zip(word_frequencies.values(), word_frequencies.keys())
    return sorted(counter, reverse=True)

# run the whole thing
word_counter = word_frequencies('holygrail.txt')
sort_dictionary_by_value(word_counter)
```

### Card Deck Simulation Extension
- Extend the `Deck` class from earlier to add a method `deal_hand(self, n)` that returns a list of `n` cards.
- Write a function that takes a list of 5 cards as a parameter and returns True if the hand is a 'Flush'.  A 'Flush' a poker hand with all five cards from the same suit.
- Write a function that deals 5 hands of 5 cards each and prints cards in each hand. Ensure the deck shrinks appropriately.  Note that you will find that the `__str__` does not get used when you print a list of objects.  There is a dunder method called `__repr__` that can do this.

```python
# Solution

import random

# Define a Card class to represent a playing card
class Card:
    def __init__(self, suit, rank):
        self.suit = suit
        self.rank = rank
    
    def __str__(self):
        return f"{self.rank} of {self.suit}"

    def __repr__(self):
        return f"{self.rank} of {self.suit}"

# Define a Deck class to represent a deck of cards
class Deck:
    def __init__(self):
        self.cards = []
        for suit in ["Hearts", "Diamonds", "Clubs", "Spades"]:
            for rank in ["Ace", "2", "3", "4", "5", "6", "7", "8", "9", "10", "Jack", "Queen", "King"]:
                self.cards.append(Card(suit, rank))
    
    def shuffle(self):
        random.shuffle(self.cards)
    
    def deal(self):
        return self.cards.pop()

    def deal_hand(self, n):
        """Return a list of n dealt from the deck."""
        hand = []
        for _ in range(n):
            hand.append(self.deal())
        return hand
        
    def __str__(self):
        return f"Deck with {len(self.cards)} cards"


def flush(hand):
    """Return True if the hand is a flush (5 cards of the same suit)."""
    # make a set of the suits in the hand 
    # if it's a flush, there will only be one unique suit
    suits = set()
    for card in hand:
        suits.add(card.suit)
        
    return len(suits) == 1

assert flush([Card("Hearts", "2"), Card("Hearts", "5"), Card("Hearts", "9"), Card("Hearts", "Jack"), Card("Hearts", "Ace")]) == True
assert flush([Card("Hearts", "2"), Card("Hearts", "5"), Card("Hearts", "9"), Card("Hearts", "Jack"), Card("Spades", "Ace")]) == False   

# Create a deck
deck = Deck()
print(deck)

print("Shuffling the deck...")
deck.shuffle()

print("Dealing 5 hands of 5 cards each...")
for i in range(5):
    
    # Deal a hand of 5 cards and print it
    hand = deck.deal_hand(5)
    print(f"Hand {i + 1}: {hand}")
    if flush(hand):
        print(f"This hand is a flush!")
    
# print the number of cards left in the deck after dealing the hands
print(f"Cards left in deck: {len(deck.cards)}\n")
```
