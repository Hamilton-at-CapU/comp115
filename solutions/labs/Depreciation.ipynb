A company's fixed assets, such as a vehicle, decline in value over time and this change is called depreciation. One way to account for this in the company books is called declining balance depreciation. Each year the value of an asset depreciates by a fixed percentage.
For example, assume that a piece of equipment is purchased for $\$$ 10,000, and it depreciates at 10% per year.  The first year depreciation of the item is: $\$$10,000 * 0.10 = $\$$1000. The item is now valued at $\$$9000. The second year depreciation of the item is: $\$$9,000 * 0.10=$\$$900. The item is now valued at \$8,100.

Write a function named `value_next_year` that takes the current value of an item (in \$) and the depreciation rate (in decimal form) as input and computes the depreciated value after one year.

```python
# Solution for your value_next_year function here
def value_next_year(current_value, dep_rate):
  '''Return the depreciated value after one year.
     Parameters
        current_value: the current value of the item
        dep_rate: the depreciation rate of the item expressed as fraction
  '''

  return current_value*(1-dep_rate)
```

Write a function named `future_value` that uses the `value_next_year` function to calculate the value of an item after `n_years`.  Input parameters should be the current value of the item (in \$), depreciation rate (in decimal form), and number of years.

```python
# Solution for your future_value function here
def future_value(current_value, dep_rate, n_years):
  '''Return the depreciated value after some number of years.
     Parameters
        current_value: the current value of the item
        dep_rate: the depreciation rate of the item expressed as fraction
        n_years: number of years
  '''

  for i in range(n_years):
    current_value = value_next_year(current_value, dep_rate)
  return current_value

```

Write a **recursive** function named `years_to_full_depreciation` that returns the number of years (integer) until a asset has a value less than $1 (ie. it is 'fully depreciated').  The function must use the `value_next_year` function and must be recursive.  Input parameters should be the current value of the item (in \$), and depreciation rate (in decimal form).

```python
# Solution for recursive number of years function here
def years_to_full_depreciation(current_value, dep_rate):
  '''Return the number of years until the value is less than a dollar (recursive version).
     Parameters
        current_value: the current value of the item
        dep_rate: the depreciation rate of the item expressed as fraction
  '''
  # Base case: if value is 1 or less, it's fully depreciated or already below 1
  if current_value <= 1:
    return 0

  # calculate the new current value (with our value_next_year function)
  new_value = value_next_year(current_value, dep_rate)

  # here's the recursive part, add 1 year to the number of years to
  # depreciate to the new value
  return 1 + years_to_full_depreciation(new_value, dep_rate)
```
