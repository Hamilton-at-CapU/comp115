## Make a Student Class

Define a class named `Student` to represent a student.  The `Student` class should have the following `attributes` and `methods`.  The attribute type is provided after the attribute name.

* Attributes
  * `name`: `str`, the student's full name
  * `courses`: `list`, a list of course names that the student is enrolled in

* Methods
  * `__init__(self, name)`: initialize the name (from parameter) and courses (empty) attributes
  * `__str__`: returns the string representation of the student name
  * `addCourses(course_list)`: adds a list of courses to the `courses` attribute
  * `isEnrolled(course_name)`: returns True iff the student is enrolled in `course_name`




````python
# Solution
class Student:
  """
    Student class represents a student.
  """

  def __init__(self, name):
    """ create a new student """
    self.name = name
    self.courses = []

  def __str__(self):
    """ string representation of a student """
    return f'Name: {self.name}'

  def getName(self):
    """ returns the name of the student """
    return self.name

  def getCourses(self):
    """ returns the list of courses the student is enrolled in """
    return self.courses

  def addCourses(self, course_list):
    """ adds a list of courses to the courses attribute """
    self.courses += course_list

  def isEnrolled(self, course_name):
    """ returns True iff the student is enrolled in course_name """
    return course_name in self.courses
````

````python
# test the student class
a_student = Student('Sir Galahad')
a_student.addCourses(['MATH116', 'STAT101', 'INTS110', 'COMP115', 'CHEM110'])
assert a_student.getName() == 'Sir Galahad'
assert a_student.getCourses() == ['MATH116', 'STAT101', 'INTS110', 'COMP115', 'CHEM110']
assert a_student.isEnrolled('COMP115') == True
assert a_student.isEnrolled('PHYS112') == False
````

## Read the Student File
Write a function named `read_student_file` that reads a student file and returns the list of students contained in the file.  The function should take the name of the file as a parameter.  An example file is `students.txt` (available on eLearn).



````python
# Solution
def read_student_file(filename):
  """
    Returns a list of students (objects of the Student class)
    Parameters: filename - (string) name of the file containing student enrolment information
  """

  # create empty list
  students = []

  # open the file for reading
  with open(filename,'r') as infile:

    # loop over lines
    for line in infile:

      # strip the newline character from the line
      line = line.strip()

      # split lines into name and courses using ';' delimiter
      name, courses = line.split('; ')

      # split courses string into a list using ',' delimiter
      course_list = courses.split(', ')

      # create the student object
      student = Student(name)

      # add the courses to the student
      student.addCourses(course_list)

      # add the student to the list of students
      students.append(student)

  # return the list of students
  return students

````

````python
# test the read_student_file function
assert len(read_student_file('students.txt')) == 12
assert [student.getName() for student in read_student_file('students.txt')] == ['King Arthur', 'Tim the Enchanter', 'The Black Knight', 'Prince Herbert', 'The Green Knight', 'The Bridekeeper', 'The Rabbit of Caerbannog', 'Brave Sir Robin', 'Roger the Shrubber', 'Sir Lancelot', 'Sir Galahad', 'Sir Bedevere']
````

## Make a Course Class

Define a class named `Course` to represent a course.  The `Course` class should have the following `attributes` and `methods`:

* Attributes
  * `name`: `str`, the course name (example, `COMP115`)
  * `students`: `list`, the students enrolled in the course

* Methods
  * `__init__(name)`: initialize the attributes
  * `__str__`: returns the string representation of the course with the course code and the number of students in the course
  * `getName()`: returns the course name
  * `addStudent(Student)`: adds a student to the student list
  * `getStudentNames()`: returns a list of student names enrolled in the course
  * `size()`: returns the number of students in the course
  * `printStudentList()`: prints the student list (note that this is different from the __str__ method because it is *printing*, not returning a string representation)


````python
# Solution
class Course:
  """ Course represents a course with students """

  def __init__(self, name):
    """ create a new course """
    self.name = name
    self.students = []

  def __str__(self):
    """ returns a string representation of a course """
    return f'Course: {self.name}, Number of Students: {len(self.students)}'

  def getName(self):
    """ returns the name of the course """
    return self.name

  def addStudent(self, student):
    """ adds a student to the student list """
    self.students.append(student)

  def getStudentNames(self):
    """ returns a list of student names enrolled in the course """
    student_names = []
    for student in self.students:
      student_names.append(student.getName())
    return student_names

  def size(self):
    """ return the number of students in the course """
    return len(self.students)

  def printStudentList(self):
    """ print the student list """
    for student in self.students:
      print(student)


````

````python
# test the Course class works
a_course = Course('COMP115')
a_course.addStudent(Student('Tim the Enchanter'))
assert a_course.getName() == 'COMP115'
assert a_course.size() == 1
assert a_course.getStudentNames() == ['Tim the Enchanter']
````

## Create the Courses Dictionary
Write a function named `create_courses` that takes the student enrollment list (ie. the output of the `read_student_file` function from part 1) and returns a courses dictionary.  The courses dictionary must have the course names as keys and Course objects as values.

````python
# Solution
def create_courses(students):
  """
    Returns a dictionary of Courses (keys: course_name, values: Course object)
    Parameters: a list of students
  """

  # create and empty dictionary of courses
  courses = {}

  # fill the dictionary of courses, excluding duplicates
  # sets are a great way to do this, but we aren't learning them this course...
  for student in students:
    for course in student.getCourses():
      if course not in courses:
        course_to_add = Course(course)
        courses[course] = course_to_add

  # return courses list
  return courses
````

````python
# test the create_courses function
courses = create_courses(read_student_file('students.txt'))
assert len(courses) == 7
assert [course.name for course in courses.values()] == ['COMP115', 'INTS110', 'MATH116', 'BIOL110', 'CHEM110', 'STAT101', 'PHYS112']
````

## Fill the Courses Dictionary
  Write a `fill_courses` function that fills each course in the course dictionary with the students enrolled in the course.  Parameters must be a list of students and a courses dictionary.  The function should return None.

````python
# Solution
def fill_courses(students, courses):
  """
    Returns None
    Parameters: - students: a list of students
                - courses: a dictionary of courses (keys: course_name, values: Course object)
  """

  # loop through the students list
  for student in students:

    # loop through the courses the student wants to enrol in
    for course_name in student.getCourses():

      # add each student to the course they are enroled in
      courses[course_name].addStudent(student)

  # return None because we are filling the dictionary object
  return None
````

````python
# test the fill_courses function
a_student = Student('Sir Galahad')
a_student.addCourses(['MATH116', 'STAT101'])
b_student = Student('Tim the Enchanter')
b_student.addCourses(['COMP115', 'STAT101'])
students = [a_student, b_student]

courses = create_courses(students)
fill_courses(students, courses)

assert len(courses) == 3
assert courses.keys() == {'MATH116', 'STAT101', 'COMP115'}
assert courses['STAT101'].size() == 2
assert courses['COMP115'].size() == 1
assert courses['STAT101'].students[0].getName() == 'Sir Galahad'
assert courses['STAT101'].students[1].getName() == 'Tim the Enchanter'
assert courses['COMP115'].students[0].getName() == 'Tim the Enchanter'
````

## Main

Write a `main` function that uses the functions and classes above to read the `students.txt` file and ask the user which class list they would like to see the list of students enrolled.

The output should produce something like this:


    Here are the available courses:
    COMP115, INTS110, MATH116, BIOL110, CHEM110, STAT101, PHYS112,

    Type a course name to get the class list: MATH116

    Course: MATH116, Number of Students: 5
    Name: King Arthur
    Name: The Green Knight
    Name: Brave Sir Robin
    Name: Sir Lancelot
    Name: Sir Galahad

````python
# Main
def main():
  """
  User interface for the class list program.
  """

  # the input filename
  infilename = 'students.txt'

  # read the file
  students = read_student_file(infilename)

  # create courses
  courses = create_courses(students)

  # fill the courses
  fill_courses(students, courses)

  # provide the user with a list of courses to choose from
  print('Here are the available courses:')
  for course in courses.keys():
    print(f'{course}, ', end='')
  print('\n')

  # ask user which course they want the student list for
  user_request = input(f'Type a course name to get the class list: ')

  # print the course information
  print()
  if user_request not in courses:
    print(f'Course {user_request} not found')
  else:
    # print the course info and student list
    print(courses[user_request])
    courses[user_request].printStudentList()
````

````python
# run main
main()

````
