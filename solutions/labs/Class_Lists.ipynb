## Make a Student Class

Define a class named `Student` to represent a student.  The `Student` class should have the following `attributes` and `methods`.  The attribute type is provided after the attribute name.

* Attributes
  * `name`: `str`, the student's full name
  * `courses`: `list`, a list of course names that the student is enrolled in

* Methods
  * `__init__(self, name)`: initialize the name (from parameter) and courses (empty) attributes
  * `__str__`: returns the string representation of the student name
  * `makeStudent(name, course_list)`: a static method that returns a Student object
  * `addCourses(course_list)`: adds a list of courses to the `courses` attribute
  * `isEnrolled(course_name)`: returns True iff the student is enrolled in `course_name`




```python
# Solution
class Student:
  """
    represents a student with enrolled courses
  """

  def __init__(self, name):
    """ create a new student """
    self.name = name
    self.courses = []

  def __str__(self):
    """ string representation of a student """
    return f'Name: {self.name}'

  def makeStudent(name, course_list):
    """ creates a student with name and course_list """
    student = Student(name)
    student.addCourses(course_list)
    return student
  
  def addCourses(self, course_list):
    """ adds a list of courses to the courses attribute """
    self.courses += course_list

  def isEnrolled(self, course_name):
    """ returns True iff the student is enrolled in course_name """
    return course_name in self.courses
```

```python
# test the student class
test_student = Student.makeStudent('Sir Galahad', ['MATH116', 'STAT101', 'INTS110', 'COMP115', 'CHEM110'])
assert test_student.name == 'Sir Galahad'
assert test_student.courses == ['MATH116', 'STAT101', 'INTS110', 'COMP115', 'CHEM110']
assert test_student.isEnrolled('COMP115') == True
assert test_student.isEnrolled('PHYS112') == False
assert str(test_student) == 'Name: Sir Galahad'
```

```python
from os.path import basename, exists

def download(url):
    filename = basename(url)
    if not exists(filename):
        from urllib.request import urlretrieve

        local, _ = urlretrieve(url, filename)
        print("Downloaded " + str(local))
    return filename

download('https://github.com/Hamilton-at-CapU/comp115/raw/main/datafiles/students.txt')
```

## Read the Student File
Write a function named `read_student_file` that reads a student file and returns a list of students contained in the file.  The function should take the name of the file as a parameter. 



```python
# Solution
def read_student_file(filename):
  """
    Returns a list of students (objects of the Student class)
    Parameters: filename - (string) name of the file containing student enrolment information
  """

  # create empty list
  students = []

  # open the file for reading
  with open(filename,'r') as infile:

    # loop over lines
    for line in infile:

      # strip the newline character from the line
      line = line.strip()

      # split lines into name and courses using ';' delimiter
      name, courses = line.split('; ')

      # split courses string into a list using ',' delimiter
      course_list = courses.split(', ')

      # use makeStudent to create the student object
      student = Student.makeStudent(name, course_list)

      # add the student to the list of students
      students.append(student)

  # return the list of students
  return students

```

## Main

Write a `main` function that uses the functions and classes above to:
   * ask the user for the name of the students file to read
   * reads the file requested by the user
   * output to the user that the file has been read and how many students there were
   * ask the user to select one of the following options
      1. print a list of student names and the number of courses the student is enrolled in
      2. print all the courses that students are enrolled in (each course should appear once)
      3. print the average number of courses per student
      4. print the average number of students per course
      5. exit
   * keep asking the user to select a choice until they select exit

```python
# Solution
def main():
    """
    Main function to handle user interaction for student data analysis.
    """
    # Ask for the filename
    filename = input("Enter the name of the students file to read: ")
    
    # Read the file
    students = read_student_file(filename)
    
    # Output the result
    print(f"File '{filename}' has been read. There are {len(students)} students.")
    
    # keep asking the user for options until they choose to exit
    while True: 
        
        # Present options
        print("\nSelect an option:")
        print("1. Print a list of student names and the number of courses each is enrolled in")
        print("2. Print all the courses that students are enrolled in (each course appears once)")
        print("3. Print the average number of courses per student")
        print("4. Print the average number of students per course")
        print("5. Exit")
        
        # Get user choice
        choice = input("Enter your choice (1-5): ")
        
        if choice == '1':
            print("\nStudent names and number of courses:")
            for student in students:
                print(f"{student.name}: {len(student.courses)} courses")
        
        elif choice == '2':
            # use a set to collect unique courses
            courses_set = set()
            for student in students:
                courses_set.update(student.courses)
            print("\nAll courses (unique):")
            
            # sort the courses alphabetically before printing
            for course in sorted(courses_set):
                print(course)
        
        elif choice == '3':
            # accumulate total number of courses for all students
            total_courses = 0
            for student in students:
                total_courses += len(student.courses)
            # calculate average courses per student
            avg_courses = total_courses / len(students)
            print(f"\nAverage number of courses per student: {avg_courses:.1f}")
        
        elif choice == '4':
            # make a counter dictionary to count how many students are in each course
            course_counts = {}
            for student in students:
                for course in student.courses:
                    if course in course_counts:
                        course_counts[course] += 1
                    else:
                        course_counts[course] = 1
                        
            # add up the total number of students in all courses
            total_students_in_courses = 0
            for count in course_counts.values():
                total_students_in_courses += count
            
            # get the number of unique courses
            num_courses = len(course_counts)
            
            # calculate average students per course
            avg_students = total_students_in_courses / num_courses 
            
            #print the average number of students per course
            print(f"\nAverage number of students per course: {avg_students:.1f}")
        
        elif choice == '5':
            print("Exiting...")
            break
        
        else:
            print("Invalid choice. Please select 1-5.")
```

```python
# run main
main()

```
