# Lab: Turtle Polygons


Write a function, `draw_polygon(n, h)`, that draws a regular (all sides are equal length) polygon with `n` sides and height `h`. Here is the math of the problem to help you.  The side length, $s$ of an $n$ sided polynomial (for $n>4$) with height, $h$ is

> $ s = \frac{h}{\tan \theta }$

where $\theta$ is:

> $ \theta = \left( \frac{1}{2}-\frac{1}{n} \right) \pi $

The above is only strictly true for polygons with an even number of sides, but is approximately true for polygons with an odd number of sides.

Write function definition here.  Don't forget the docstring.

```python
# Solution

def draw_polygon(n, h):
    """
    Draws a polygon using a jupyturtle
    n: integer number of sides
    h: integer height of the polygon
    """

    # make a turtle
    make_turtle()

    # calculate the angle
    angle = (math.pi)*(1/2-1/n)

    # calculate the side_length using the angle
    side_length = height / math.tan(angle)

    # loop over the number of sides
    for i in range(n):

      # draw the side
      forward(side_length)

      # turn left by 360/number of sides
      left(360/n)
```

Call the function to draw a polygon here.

```python
# Solution here

draw_polygon(n_sides, height)
```
