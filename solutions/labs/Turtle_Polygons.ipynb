Write a program that draws a regular (all sides are equal length) polygon with `n` sides and height `h`. The program should include generalized functions that encapsulate the various parts of the problem.  Here is the math of the problem to help you.  The side length, $s$ of an $n$ sided polynomial (for $n>4$) with height, $h$ is

> $ s = \frac{h}{\tan \theta }$

where $\theta$ is:

> $ \theta = \left( \frac{1}{2}-\frac{1}{n} \right) \pi $

The above is only strictly true for polygons with an even number of sides, but is approximately true for polygons with an odd number of sides.

Write the function definitions here.  Don't forget the docstrings.

```python
# Solution

# a function to get the side length of a polygon
# using the equation provided
def side_length(n,h):
  """
  returns the side length of an n-sided polygon with height h
  n: integer number of sides
  h: integer height
  """
  # theta from the equations provided
  theta = (math.pi)*(1/2-1/n)

  # calculate and return the side length using the equations provided
  return h / math.tan(theta)

# a draw polygon function using the jupyturtle module
def draw_polygon(n,h):
  """
  draws an n-sided polygon with height h
  n: integer number of sides
  h: integer height
  """

  # make a turtle
  make_turtle()

  # loop over the number of sides
  for _ in range(n):

      # move forward by side length
      forward( side_length(n,h) )

      # turn left by external angle (360//n)
      left( 360 // n )
```

Call the function(s) to draw a polygon here.

```python
# Solution

draw_polygon(n_sides, height)
```
