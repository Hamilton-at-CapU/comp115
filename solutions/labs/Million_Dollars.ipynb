# Lab: Million Dollars 
*[If I had a million dollars...](https://www.youtube.com/watch?v=aynCgnbbgbM)*.  In this lab you will calculate what fraction of your salary each month you would need to save in order to have $1,000,000 by the time you retire.  The user will be asked for their current annual salary, the current value of their savings, and the number of years until they retire.  The program will tell the user what fraction of their salary they need to save each month to achieve their retirement dreams.

For example,
> `How much do you currently have saved?`  1000

> `What is your current annual salary?`  60000

> `How many years until you retire?`  40

> `You need to save 6.58% of your salary to have $1000000 for retirement.`

We will make a couple broad assumptions in order to make the result at least somewhat plausible.  Assume that the saving account pays you interest with an annual rate of return (`r`) of 6% (i.e. `r=0.06`), compounded monthly, so that at the end of each month you receive `savings*(r/12)` to put back into your savings (note `r` is an annual rate, so `r/12` is the monthly rate). In other words, each month, your savings will be increased by the return on your investment plus the fraction of your monthly salary that you decide to save each month. Hopefully, you won't have the same salary for the rest of your life.  So, let's also assume that you start at some salary and that you will get a 3% raise every year until retirement.  





## Write some small helper functions

The following functions are designed to *decompose* the problem into a collection of *abstract* functions with *low coupling* that can be put together to solve the larger problem in the `main` code.  These functions are designed to solved the problem by *iterating* over each month and calculating the amount of savings *accumulated* over your career.

- **`monthly_contribution(salary, portion_saved)`** Returns the amount to be contributed to savings from the monthly salary where `salary` is the current monthly salary and `portion_saved` is the fraction of salary saved.  Example; if the monthly salary is \$6,000 and the portion saved is 10%, then the function should return \$600.

```python
# Solution for monthly_contribution function here
def monthly_contribution(monthly_salary, portion_saved):
  '''
  Return the amount to be contributed to savings from the monthly salary.
  Parameters:
    monthly_salary: the current monthly salary
    portion_saved: the fraction of salary saved
  '''

  return monthly_salary*portion_saved

assert monthly_contribution(0,0) == 0
assert monthly_contribution(0,1) == 0
assert monthly_contribution(10,0) == 0
assert monthly_contribution(10,0.5) == 5
assert monthly_contribution(1000,0.1) == 100
assert monthly_contribution(6000,0.1) == 600
```

- **`monthly_interest(savings, annual_interest_rate)`** Returns the amount of interest accrued for the month on the current savings amount.  Example; if the savings account has \$100,000 and the annual interest rate is 12%, then the function will return \$1,000.



```python
# Solution for monthly_interest function here
def monthly_interest(savings, annual_interest_rate):
  '''
  Return the amount of interest accrued for the month.
  Parameters:
    savings: the current value of the savings account
    annual_interest_rate: the annual interest rate
  '''

  return savings * (annual_interest_rate/12)

assert monthly_interest(0,0) == 0
assert monthly_interest(1,0) == 0
assert monthly_interest(0,1) == 0
assert monthly_interest(1,12) == 1
assert monthly_interest(100,0.12) == 1
assert monthly_interest(1000,0.06) == 5
assert round(monthly_interest(100000,0.1),2) == 833.33
assert round(monthly_interest(250010,0.06),2) == 1250.05
```

- **`is_anniversary(month_number)`** Returns True if the number of months since starting work (`month_number`) is an anniversary and thus you should get a raise.  Example; if you are in the 15th month of work the function should return False, if you are in your 36th month of work the function should return True.  Note that the first month of work should not considered an anniversary because you don't (typically) get a raise in your first month.

```python
# Solution for is_anniversary function here
def is_anniversary(month_number):
  '''
  Returns True iff this month is an anniversary.  The first day of work is
  not considered an anniversary.
    Paramaters:
      month_number: the number of months since starting work
  '''

  return month_number!=0 and month_number%12==0

assert is_anniversary(0) == False
assert is_anniversary(1) == False
assert is_anniversary(15) == False
assert is_anniversary(2.5) == False
assert is_anniversary(12) == True
assert is_anniversary(24) == True
assert is_anniversary(36) == True
```

- **`apply_raise(salary, annual_raise)`** Returns the new salary after the annual raise has been applied to the current salary.  Example; a salary of \$1,000 with a raise of 3% the function should return \$1,030

```python
# Solution for apply_raise function here
def apply_raise(salary, annual_raise):
  '''
  Returns the new salary after the annual raise has been applied.
    Parameters:
      salary: the current salary
      annual_raise: the annual raise to apply
  '''

  return salary*(1+annual_raise)

assert apply_raise(0,0) == 0
assert apply_raise(1,0) == 1
assert apply_raise(0,1) == 0
assert apply_raise(1,1) == 2
assert apply_raise(10,0.03) == 10.30
assert apply_raise(1000,0.03) == 1030
assert round(apply_raise(10001,0.03),2) == 10301.03
```

## Write an `amount_saved` function

Using the helper functions written above, write a function that returns the amount saved by the time you retire.  

```python
# Solution for amount_saved function 
def amount_saved(initial_savings,
                 monthly_salary,
                 portion_saved,
                 annual_raise,
                 return_rate,
                 months_to_retirement):
  """
  Returns the amount of money saved at the time of retirment.

  Parameters:
    initial_savings: the initial amount of money saved
    monthly_salary: monthly salary (in dollars)
    portion_saved: fraction of monthly salary that is saved (as a fraction)
    annual_raise: salary increase every 12 months (as a fraction)
    return_rate: the annual rate of return on the invested savings (as a fraction)
    months_to_retirment: the number of months until retirement

  """

  # initiate savings
  savings = initial_savings

  # increment savings each month until retirement
  for n in range(1,months_to_retirement+1):

    # apply raise if it is an anniversary
    if is_anniversary(n):
      monthly_salary = apply_raise(monthly_salary, annual_raise)

    # accumulate the savings for the month
    savings = savings \
              + monthly_contribution(monthly_salary, portion_saved) \
              + monthly_interest(savings, return_rate)

  # return the amount saved
  return savings
```

## Write some I/O helper functions

Write the following functions to help collect/validate user input and to format output to the user. - **`get_annual_salary()`** Gets the annual salary from the user and returns the annual salary as a float.  You may assume that the user inputs a number, but you must check that it is a positive number and keep asking until the user enters a positive number.

```python
# Solution for get_annual_salary function here
def get_annual_salary():
  ''' return annual salary from user input'''

  user_input = float(input("What is your current annual salary? "))
  while user_input <= 0:
    print("Please enter a positive number.")
    user_input = float(input("What is your current annual salary? "))
  return user_input
```

- **`get_initial_savings()`** Asks the user 'Do you have have any current savings (y/n)? '  If they respond 'y', ask them what their current savings is and return the current savings as a float.  You may assume that the user inputs a number.  If they reply 'n', then return 0.  If they reply with anything other than 'y' or 'n', ask them again.

```python
# Solution for get_initial_savings function here
def get_initial_savings():
  ''' return initial savings from user input as a float'''

  while True:
    user_response = input("Do you have any current savings (y/n)? ")
    if user_response == 'y':
      user_input = float(input("How much do you currently have saved? "))
      return user_input
    if user_response == 'n':
      return 0
    print("Please enter 'y' or 'n'.")
```

- **`get_months_to_retirement()`** Asks the user for the number of years until retirement and returns the number of months until retirement as an int.  You may assume that the user inputs a number, but do not assume the user inputs an integer.  They could enter 4.5, meaning 4 years and 6 months to retirement.  

```python
# Solution for get_months_to_retirement function here
def get_months_to_retirement():
  ''' return number of months to retirement from user input as number years'''

  user_input = float(input("How many years to retirement? "))
  return int(user_input*12)
```

- **`print_result(fraction, target)`** Print the result with the correct formatting where `fraction` is the fraction of the users salary they need to save to reach the `target` amount.  The correct formatting is given in the example above.  Note the output formatting requires the fraction given as a percent.

```python
# Solution for print_result function here
def print_result(fraction, target):
  '''
  Print the result with the correct formatting, which is given in the example.

  Parameters:
    fraction: the fraction of the users salary they need to save to reach the target
    target: the amount of money you want to have at retirement
  '''
  percentage = 100*fraction
  print(f'You need to save {percentage:.3}% of your salary to have ${int(target)} for retirement.')

```

## Put it all together
In the `amount_saved` function you calculated how much you will have saved for retirement after a fixed number of years at a fixed fraction saved.  Use the `amount_saved` function to write an algorithm than determines what fraction of your salary you need to save in order have a million dollars at retirement.  In other words, the amount saved is now fixed (one million dollars) and the fraction saved is now the unknown variable.  

The assumed values:
* Savings have an annual return on investement of 6%.
* You get a 3% raise every year.
* Your target savings amount is $1,000,000.

Input values from the user are:
* Current amount of savings.
* Current annual salary
* The number of years until retirement.

Output:
* The percentage of salary required to save in the format given in the example at the begining of the lab.



Initialize the following constants to represent these assumed values

 - RATE_OF_RETURN, annual rate of return on investment
 - ANNUAL_RAISE, the increase in your salary each year
 - TARGET_SAVINGS, the amount of money you want to have at retirement


```python
# define contants here
RATE_OF_RETURN = 0.06      # annual rate of return on investment
ANNUAL_RAISE = 0.03        # the increase in your salary each year
TARGET_SAVINGS = 1e6       # the amount of money you want to have at retirement
```

```python
# wSolution for the main() function here
def main():

  # collect the user input
  initial_savings = get_initial_savings()
  annual_salary = get_annual_salary()
  months_to_retirement = get_months_to_retirement()

  # calculate the initial monthly salary
  initial_monthly_salary = annual_salary/12

  # test if it is possible to save enought assuming 100% of your salary is saved
  max_savings = amount_saved(initial_savings,
                             initial_monthly_salary,
                             1.0,
                             ANNUAL_RAISE,
                             RATE_OF_RETURN,
                             months_to_retirement)

  # if it is not possible to save enough money, tell the user and exit the function
  # note: the 'proper' way to do this in Python is by 'raising' and 'exception',
  # but we won't be learning that in this course
  if max_savings < TARGET_SAVINGS:
    print('It is not possible to save that much money. \
            You need a higher paying job, or lower retirement expectations!')
    return -1


  # The following code uses the 'bisection search algorithm' to determine
  # the fraction that will need to be saved.  Note that this is not the
  # only way to solve this problem.

  # initialize a savings_at_retirment variable to compare to the target savings
  savings_at_retirement = 0

  # set the limits of the bisection search algorithm
  min = 0.0     # lower bound (0% of salary saved)
  max = 1.0     # upper bound (100% of salary saved)

  # set the threshold of success for bisection search
  epsilon = 1000  # savings must be within this amount of the target savings

  # initialize a counter for the number of guesses
  num_guess = 0

  # use the bisection method to iterate over different portion_saved values (ie. guesses)
  # exit the bisection search when current savings is within epsilon of the target_savings
  while abs(savings_at_retirement - TARGET_SAVINGS) > epsilon:

    # set the guess of the bisection method to be half way between the
    # upper and lower bounds
    guess = (min+max)/2

    # set the portion_saved (a float) from the guess (an integer)
    portion_saved = guess

    # calculate the savings at retirement for the protion_saved in this loop
    savings_at_retirement = amount_saved(initial_savings,
                                         initial_monthly_salary,
                                         portion_saved,
                                         ANNUAL_RAISE,
                                         RATE_OF_RETURN,
                                         months_to_retirement)


    # check if the guessed portion_saved saves more money than the required down payment
    # if yes, reset the upper bound of the bisection search limit to the guess
    if savings_at_retirement > TARGET_SAVINGS:
      max = guess

    # if not, reset the lower bound of the bisection search limit to the guess
    else:
      min = guess

    # increment the number of guesses
    num_guess += 1

  # need to round the output....
  print_result(guess, TARGET_SAVINGS)

  # optional print just to see how many iterations my algorithm took
  print(f'(Steps in bisection search: {num_guess})')

main()
```
